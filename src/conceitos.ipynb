```python
É uma fila ordenada de coisas. Ao contrário de uma variável normal (que só guarda uma coisa de cada vez), uma lista guarda dezenas ou milhares de coisas seguidas, numeradas a partir do 0
```

```python
// 1. Criar a lista vazia
List<String> compras = new ArrayList<>();

// 2. Adicionar frases lá dentro
compras.add("Pão");
compras.add("Leite");
compras.add("Maçãs");
```

diferença para o Array comum:

Array comum (String[])
Tamanho	Fixo. Se criares para 3 coisas, nunca cabem 4.
Adicionar itens	Tens de saber a posição exata: array[0] = "Pão";
Apagar itens Muito difícil (fica um espaço vazio no meio).
Ver o tamanho	array.length

lista (List<String>)
Flexível. Podes adicionar ou apagar à vontade.
É só mandar lá para dentro: lista.add("Pão");
Apagar itens Super fácil: lista.remove("Pão"); (os outros chegam-se à frente).
Ver o tamanho lista.size()


Quando usar cada um?

• Use Array se o tamanho for fixo e a performance ou o uso de memória for crítico.
• Use ArrayList na maioria dos casos em que precisa de uma lista ordenada que cresce dinamicamente e permite repetições.
• Use Set sempre que precisar de eliminar duplicados automaticamente ou verificar rapidamente se um item existe na coleção.

Em Java, o Map é uma estrutura de dados que armazena informações em pares de Chave e Valor (Key-Value), funcionando de forma semelhante a um dicionário ou a uma tabela de consulta. Ao contrário do Array, ArrayList e Set, o Map não guarda itens isolados, mas sim associações onde cada chave única aponta para um valor específico

```python
import java.util.HashMap;
import java.util.Map;

public class ExemploMap {
    public static void main(String[] args) {
        // Criação: Chave do tipo String (produto) e Valor do tipo Double (preço)
        Map<String, Double> precos = new HashMap<>();

        // Adicionar elementos usando .put()
        precos.put("Café", 1.20);
        precos.put("Pão", 0.50);
        precos.put("Leite", 1.10);

        // Atualizar valor (como a chave "Café" já existe, o preço é substituído)
        precos.put("Café", 1.35);

        // Procurar valor pela chave usando .get()
        System.out.println(precos.get("Café")); // Saída: 1.35

        // Verificar se uma chave existe
        System.out.println(precos.containsKey("Pão")); // Saída: true

        // Percorrer chaves e valores
        for (Map.Entry<String, Double> item : precos.entrySet()) {
            System.out.println(item.getKey() + ": " + item.getValue() + "€");
        }
    }
}
```
